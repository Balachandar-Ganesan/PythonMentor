```python
def print_array(arr, i=0):
    if i == len(arr):
        return

    print(arr[i])
    print_array(arr, i + 1)
```

```python
class Engine:
    def start(self):
        print("Engine started.")

class Car:
    def __init__(self, engine):
        self.engine = engine  # Aggregation

    def start(self):
        self.engine.start()
```

```python
AggregationEngine=Engine()
AggregationCar=Car(AggregationEngine)
AggregationCar.start()
```

```python
def duplicates1(L):
 n = len(L)
 for i in range(n):
   for j in range(n):
     if i != j and L[i] == L[j]:
       return True
   return False
assert( not duplicates1([1,2,3,4]))
print('pass')
assert(duplicates1([1,2,6,3,4,5,6,7,8]))
```

```python
import time
def timetrials(func, n, trials = 10):
 totaltime = 0
#start = time.time()
 for i in range(trials):
   start = time.time() # it should be here
   func(list(range(n)))
   totaltime += time.time()- start
   print("average =%10.7f for n = %d" % (totaltime/trials, n))
```

```python
class Node(object):
    def __init__(self, x, children=[]):
        self.x = x
        self.children = children
```

```python
def printTree(root, level=0):
    print("  " * level, root.x)
    for child in root.children:
        printTree(child, level + 1)

tree = Node("1", [Node("2", [Node("3"), Node("4")]), Node("5")])
printTree(tree)
```
