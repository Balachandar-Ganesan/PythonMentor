```python
def print_array(arr):

    for i in range(len(arr)):

        print(arr[i])
```

```python
arr=[1,3,2,5,4]
print_array(arr)
```

```python
def print_array(arr, i=0):
    if i == len(arr):
        return

    print(arr[i])
    print_array(arr, i + 1)
```

```python
arr=[1,3,2,5,4]
print_array(arr)
```

```python
def print_array(arr, i=0):
    if i == len(arr):
        return

    print_array(arr, i + 1)
    print(arr[i])
```

```python
arr=[1,3,2,5,4]
print_array(arr)
```
